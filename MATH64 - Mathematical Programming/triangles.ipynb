```python
def sum_path(nextrow, prevsol):
    result = nextrow.copy()

    result[0] += prevsol[0]
    result[-1] += prevsol[-1]

    for i in range(len(prevsol)-1):
        left = prevsol[i]
        right = prevsol[i+1]
        m = max([left, right])
        result[i+1] += m

    return result
```
