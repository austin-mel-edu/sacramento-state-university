## 2.2
### Describe the difference in roles assumed by the validation partition and the holdout partition.
The validation partition is used to assess the actual predictive performance of each model. This allows you to compare models and select the best one. The holdout partition, also called the test partition, is used to assess the performance of the chosen model with new data. We have both because when we use the validation data to assess multiple models and then choose the model that performs best with the validation data, the chosen model may be overfitted to the given data. Applying the model to the holdout data, which the model has not seen before, will provide an unbiased estimate of the true model's performance.

## 2.5
### Using the concept of overfitting, explain why when a model is fit to training data, zero error with those data is not necessarily good.
When a model is fit to training data, sometimes the model will produce a result with zero error. This occurs when the model has overfit to the data, meaning it performs perfectly on the data it was trained with, but generalizes poorly and the model performs poorly on new or unseen data. This leads to a model with higher test error and a significantly reduced predictive accuracy and model value resulting in unreliable predictions for future observations.
