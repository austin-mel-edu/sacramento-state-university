# QUESTION 1
## Convergence of the Sequence
We want to find the **smallest integer** \( n \) such that:

$$
|e - x_n| < 10^{-7}, \quad \text{where } x_n = (1 + \frac{1}{n})^n
$$

```python
n = 1
x_n = (1 + 1/n) ** n
while abs(math.e - x_n) >= 1e-7:
    n += 1
    x_n = (1 + 1/n) ** n

print(f"Smallest n: {n}")
# 13076876
```

# QUESTION 2
## Density of sin(n) in [-1, 1]

We define a function that finds the smallest natural number $n$ such that:

$$
|x - \sin(n)| < d
$$

This demonstrates that the values of $\sin(n)$ are dense in the interval $[-1, 1]$.

```python
def sin_close_int(x, d):
    n = 1
    while True:
        if abs(x - math.sin(n)) < d:
            return n
        n += 1

print(sin_close_int(0.5, 0.001))
# 3783
```

# QUESTION 4
## Plotting Functions f(x), g(x), and h(x)

We will plot:
- $f(x) = x^3 - 2x^2 - x + 2$ for $x \in [-2, 3]$
- $g(x) = \cos^2(x)$ for $x \in [0, 6\pi]$
- $h(x) = e^{-x^2}$ for $x \in [-5, 5]$

```python
x1 = np.linspace(-2, 3, 200)
f = x1**3 - 2*x1**2 - x1 + 2

x2 = np.linspace(0, 6*np.pi, 400)
g = np.cos(x2)**2

x3 = np.linspace(-5, 5, 400)
h = np.exp(-x3**2)

fig, axs = plt.subplots(3, 1, figsize=(7, 10))

axs[0].plot(x1, f)
axs[0].axhline(0, color='black', lw=0.8)
axs[0].set_title(r"$f(x) = x^3 - 2x^2 - x + 2$")

axs[1].plot(x2, g)
axs[1].axhline(0, color='black', lw=0.8)
axs[1].set_title(r"$g(x) = \cos^2(x)$")

axs[2].plot(x3, h)
axs[2].axhline(0, color='black', lw=0.8)
axs[2].set_title(r"$h(x) = e^{-x^2}$")

plt.tight_layout()
plt.show()
```

# QUESTION 5
## Plotting sin(x) function with Default Argument

We define a function `plotsin(n=20)` that plots $\sin(x)$ for $n$ equally spaced points between $0$ and $2\pi$.

```python
def plotsin(n=20):
    x = np.linspace(0, 2*np.pi, n)
    y = np.sin(x)

    plt.figure(figsize=(6,4))
    plt.plot(x, y, 'o-', label='sin(x)')
    plt.axhline(0, color='black', lw=0.8)
    plt.title(r"$\sin(x)$ with n=" + str(n))
    plt.xlabel("x")
    plt.ylabel("sin(x)")
    plt.legend()
    plt.show()

plotsin()
plotsin(100)
```
